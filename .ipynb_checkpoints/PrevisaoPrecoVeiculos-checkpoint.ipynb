## Projeto 1 - Prevendo valores de veículos

## Definição do Problema de Negócio

O objetivo é construir um modelo de Machine Learning que seja capaz de fazer previsões sobre o valor de venda de veículos, avaiando o melhor score apresentado.
Para resolver esse problema será utilizado Regressão Linear.

**Fonte dos dados** - kaggle - https://www.kaggle.com/datasets/athirags/car-data

**Informações variáveis:**

Car_Name: Nome do carro  
Year: Ano que o carro foi vendido  
Present_Price: Preço atual decidido  
Mks_Driven: Total de km rodados  
Fuel_Type: Tipo de combustível (Diesel ou Petrol(Gasolina))  
Seller_Type: Tipo de venda  
Transmission: Tipo de Cambio (manual ou automático)  
Owner: Dono  
Selling_Price: O preço pelo qual o carro foi comprado(variável TARGET)

## Pré Processamento dos Dados / Tratamento dos Dados

```python
#Obtendo a data atual
datetime = datetime.datetime.now()
print(datetime)
```

```python
#Verificando o ano atual
datetime.year
```

```python
#Engenharia de atributo - obtendo a idade do carro
df['Age'] = datetime.year - df['Year']
```

```python
#Ao verificar novamente o Dataframe há uma nova coluna criada a partir de uma coluna atual
df.head()
```

```python
#Excluir a variavel ano
df.drop('Year', axis = 1, inplace = True)
df.head()
```

```python
#Identificado na analise do boxplot um possivel outlier para a variavel Selling_Price
#Aplicar um filtro para avaliar os dados e remover o outlier

sorted(df['Selling_Price'], reverse = True)
```

```python
#Avaliar qntas observacoes existem com valor maior ou igual a 33
df[(df['Selling_Price']>=33.0)]
```

```python
#Avaliar os dados existem com valor menor a 33
df[(df['Selling_Price']<33.0)]
```

```python
#Remocao das 2 linhas criando um novo dataset com o filtro aplicado
#Novo Dataframe com o mesmo nome df porém aplicando já o filtro para considerar tudo o que for inferior a 33

df = df[(df['Selling_Price']<33.0)]
```

```python
df.shape
```

```python
#Transformar variaveis strings em numeros (Encoding)
df['Fuel_Type'].unique()
```

```python
df.loc[:, 'Fuel_Type'] = df['Fuel_Type'].map({'Petrol':0, 'Diesel':1, 'CNG':2})
```

```python
df.head()
```

```python
#Transformar variaveis strings em numeros (Encoding)
df['Seller_Type'].unique()
```

```python
df.loc[:, 'Seller_Type'] = df['Seller_Type'].map({'Dealer':0, 'Individual':1})
```

```python
#Transformar variaveis strings em numeros (Encoding)
df['Transmission'].unique()
```

```python
df.loc[:, 'Transmission'] = df['Transmission'].map({'Manual':0, 'Automatic':1})
```

```python
df.head(20)
```

```python
#Excluir as variaveis que são do tipo object que não serão utilizadas
X = df.drop(['Car_Name', 'Selling_Price'], axis = 1)   #variaveis preditoras
Y= df['Selling_Price']   #variavel Target
```

```python
#Visualizar variaveis preditoras
X.head()
```

```python
#Visualizar a variavel target
Y.head()
```

```python
print('Quantidade de observações de variáveis preditoras: ', X.shape[0])
print('Quantidade de observações da variável alvo: ', Y.shape[0])
```

## Separação dos Dados em Treino e Teste

```python
#Seperar os dados em treino e teste

#Funcao train_test_split do pacote sklearn
#X_Treino: variaveis preditoras para treino com 75% do conjunto de dados
#Y_Teste: variaveis preditoras para teste com 25% do conjunto de dados
#Y_Treino: variavel alvo com 75% do conjunto de dados
#Y_Teste: variavel alvo com 25% do conjunto de dados

X_Treino, X_Teste, Y_Treino, Y_Teste = train_test_split(X, Y, test_size = 0.25, random_state = 42) 
```

```python
print('Quantidade de observações de variáveis preditoras de treino: ', X_Treino.shape[0])
print('Quantidade de observações de variáveis preditoras de teste: ', X_Teste.shape[0])
print('Quantidade de observações de variável alvo de treino: ', Y_Treino.shape[0])
print('Quantidade de observações de variável alvo de teste: ', Y_Teste.shape[0])
```

## Escolha do Algoritmo e Criação e Treino do Modelo

```python
#Criação do algoritmo
lr = LinearRegression()

#Treinando o algoritmo
lr.fit(X_Treino, Y_Treino)
```

```python
#Criação do algoritmo
rf = RandomForestRegressor()

#Treinando o algoritmo
rf.fit(X_Treino, Y_Treino)
```

```python
#Criação do algoritmo
xgb = GradientBoostingRegressor()

#Treinando o algoritmo
xgb.fit(X_Treino, Y_Treino)
```

```python
X_Treino.dtypes  #Consultar se os dados das colunas são numericos
```

```python
#Conversao de texto para numerico

cols = ['Fuel_Type', 'Seller_Type', 'Transmission']

for c in cols:
    X_Treino[c] = pd.to_numeric(X_Treino[c], errors='coerce')
    X_Teste[c] = pd.to_numeric(X_Teste[c], errors='coerce')
```

```python
print(X_Treino.dtypes)
```

```python
#Criação do algoritmo
xg = XGBRegressor()

#Treinando o algoritmo
xg.fit(X_Treino, Y_Treino)
```

```python
#Fazendo as previsões
Previsoes_LR = lr.predict(X_Teste)
Previsoes_RF = rf.predict(X_Teste)
Previsoes_XGB = xgb.predict(X_Teste)
Previsoes_XG = xg.predict(X_Teste)
```

```python
#Verificando as previsões
Previsoes_LR
```

```python
#Verificando as previsões
Previsoes_RF
```

```python
#Verificando as previsões
Previsoes_XGB
```

```python
#Verificando as previsões
Previsoes_XG
```

## Avaliação da Performance dos Algoritmos (Escala Real)

```python
#Avaliar a performance dos algoritmos
score_LR = metrics.r2_score(Y_Teste, Previsoes_LR)
score_RF = metrics.r2_score(Y_Teste, Previsoes_RF)
score_XGB = metrics.r2_score(Y_Teste, Previsoes_XGB)
score_XG = metrics.r2_score(Y_Teste, Previsoes_XG)
```

```python
print('Score utilizando Linear Regression: ', score_LR)
print('Score utilizando Linear Random Forest Regressor: ', score_RF)
print('Score utilizando GradientBoostingRegressor: ', score_XGB)
print('Score utilizando XGBRegressor: ', score_XG)
```

```python
#DataFrame com os resultados das avaliações dos algoritmos
final_data = pd.DataFrame({'Models': ['LR', 'RF', 'GBR', 'XG'], 'R2_Score': [score_LR, score_RF, score_XGB, score_XG]})
final_data
```
